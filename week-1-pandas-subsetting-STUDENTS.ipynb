## Selecting a single column

Simplest case: Selecting a single column by column name.

General syntax:

'df['column_name']'

notice column name is given a *string* inside the square brackets.
This is an example of **label-based subsetting** which means we select data from the df using the *names* of the columns, not their *position*.


#### Example 
Select the number of bird species observed at the Mugu Lagoon in spring.

````python
# Select a single column using square brackets
mul_spring = df['MUL_spring']
mul_spring # every column is a panda series
````

Since we only selected a single column, `mul_spring` is a `pandas.Series`:

````python
# Confirm type of the output
type(mul_spring)
````

Recall: We can think of a `pandas.DataFrame` as a dictionary of its columns.
Then, we can access a single column using the column name as the key, just like we would do in a dictionary: `df['column_name']`.

We can also fo label-based subsetting using synatx:
'df.column_name'

For example:

````python
df.MUL_spring
````




In general, it is better to use the `df['column_name']` syntax. Why?


### ... using a slice

To select a slice of the columns we will use a special case of **`loc` selection**. General syntax:

```python
df.loc[ : , 'column_start':'column_end']
```

'column_start' and 'column_end' = starting and endpoint of the column slice we want to subset from the dataframe.

Notice:

- the first value passed to `loc` is used for selecting rows, using a colon `:` as the row-selection parameter means "select all the rows"

- the slice of the data frame we'll obtain includes both endpoints of the slice `'column_start':'column_end'`. 


#### Example

Select the slice of columns that covers data from CSM and Mugu Lagoon.

````python
csm_mul = df.loc[ : , 'CSM_winter':'MUL_fall']
csm_mul.head()
````

````python
# Select all columns that have fall in it
df.filter(like='fall')
````

## Selecting rows...
Let's move on to selecting rows. 

### ... using a condition
To select rows that satisfy a particular condition the general syntax is 


```python
df[condition_on_rows]
```
That `condition_on_row` can be many things, let's see some usual scenarios.

#### Example

We are interested in all data after 2020.

````python
# Select all rows with year > 2020
post_2020 = df[df['year']>2020]
post_2020
````

*Let's break down what is happening here.*

condtion for our rows = `df['year']>2020` this is a `panda.Series` with boolean values (`Ttrue` or `False`) indicating which rows statisfy the condtion year>2020

````python
# Check the type of df['year']>2020
df['year']>2020
````

When we pass such a series of boolean values to the selection brackets `[]` we keep only the rows that correspond to a `True` value.

### ... using multiple conditions
We can combine multiple conditions to select rows by surrounding each one in parentheses `()` and using the or operator `|` and the and operator `&`.

#### Example: or 

Select rows in which the CSM has more than 50 species registered in winter *or* fall:

````python
df[(df['CSM_winter']>50)|(df['CSM_fall']>50)]
````

#### Example: and

Select rows in which both the Carpinteria Salt Marsh and the San Dieguito Wetland have more than 60 reported bird species during spring:

````python
df[(df['CSM_spring']>60)&(df['SDW_spring']>60)] # no selection of year
````


### ... by position

Sometimes we may want to select certain rows depending on their *actual position* in the data frame. In other words, using **position-based subsetting**.  

syntax:
```python
df.iloc[row-indices]
```

`iloc` stands for integer-location based indexing

#### Example

````python
# Selecting 5th row
df.iloc[4:5]
````

````python
df.iloc[[4]]
````

````python
# Select the fifth row (index=4)
df.iloc[4]
````

````python
# Select rows 9 through 13, including 14
df.iloc[9:14]
````

### Example

Our `post_2020` subset kept the original row labels from `df`: 11, 12, and 13.

````python
post_2020
````

````python
post_2020.iloc[[0]]
````

````python
post_2020.loc[[11]]
````

Both return the 2021 row. `iloc` counts positions starting from 0, while `loc` uses the labels in the index. That's why `post_2020.loc[0]` raises a `KeyError`: there is no row labeled 0 in `post_2020`.

## Selecting rows and columns simultaneously...

Can be done using `loc` (labels) or `iloc` (positions).

### ...by labels or conditions
When we want to select rows and columns simultaneously by labels (including using conditions) we can use `loc` selection with the syntax 

```python
df.loc[row-selection, column-selection]
```


#### Example

Select winter surveys for Mugu Lagoon and Tijuana Estuary after 2020:

````python
df.loc[df['year']>2020,['MUL_winter','TJE_winter']]
````

Let's break down this code:

- using the `df.loc[ row-selection , column-selection]` syntax

- `row-selection` parameter = condition `df['year']>2020`, which is a boolean array saying which years are greater than 2020

- `column-selection` parameter = `['MUL_winter','TJE_winter']`, which is a list with the names of the two columns we are interested in. 


### ... by position

When we want to select rows and columns simultaneously by position we use `iloc` selection with the syntax:

```python
df.iloc[row-indices, column-indices]
```
#### Example

Suppose we want to select rows 3-7 (including 7) and columns 3 and 4:

````python
df.iloc[3:8,[3,4]]
````

Let's break it down:

- we are using the `df.iloc[ row-indices , column-indices]` syntax to select by position

- the `row-indices` parameter is the slice *of integer indices* 3:8. Remember the right endpoint (8) won't be included.

- the `column-indices` parameter is the list of integer indices 3 and 4. This means we are selecting the fourth and fifth column.

## Notes about `loc` and `iloc`

## `iloc` vs. `loc`: which one does what?
*At the beginning, the difference between `iloc` and `loc` can be confusing.*

Remember the `i` in `iloc` stands for *integer location*, so `iloc` only uses integer indexing to retrieve information from the data frames in the same way as indexing for Python lists.

## `iloc` for column selection? Think about it.
We can also access columns by position using `iloc` - but it is best not to if possible.

Example:

Suppose we want to access the 9th column in the df - then we want to select a column *by position*. In this case, the 9th column is the data at San Dieguito Wetland during spring and the 9th position corresponds to the index 8.

````python
# Select column by position using iloc
# The syntax is iloc[row-indices, column-indices]
# [:,8] means "select all rows from the column at index 8"
sdw_spring = df.iloc[:,8]
sdw_spring.head()
````

Unless you are *really* looking for information about the "first column", **do not access a column by position**. This can break in many ways:

- it relies on a person correctly counting the position of a column. Even with a small dataset this can be prone to error.

- it is not explicit: if we want information about SDW during spring, `df.SDW_spring` or `df['SDW_spring']` are explicitly telling us we are accessing that information. `df.iloc[:,8]` is obscure and uninformative.

- datasets can get updated. Maybe a new column was added before `SDW_spring`, changing the position of the column and making any code depending on `df.iloc[:,8]` invalid. 

**Accessing columns by labels helps reproducibility!**
