## Selecting a single column

Simplest case: Selecting a single column by column name.

General syntax:

'df['column_name']'

notice column name is given a *string* inside the square brackets.
This is an example of **label-based subsetting** which means we select data from the df using the *names* of the columns, not their *position*.


#### Example 
Select the number of bird species observed at the Mugu Lagoon in spring.

```python
# Select a single column using square brackets
mul_spring = df['MUL_spring']
mul_spring # every column is a panda series
```

Since we only selected a single column, `mul_spring` is a `pandas.Series`:

```python
# Confirm type of the output
type(mul_spring)
```

Recall: We can think of a `pandas.DataFrame` as a dictionary of its columns.
Then, we can access a single column using the column name as the key, just like we would do in a dictionary: `df['column_name']`.

We can also fo label-based subsetting using synatx:
'df.column_name'

## Notes about `loc` and `iloc`

## `iloc` vs. `loc`: which one does what?
*At the beginning, the difference between `iloc` and `loc` can be confusing.*

Remember the `i` in `iloc` stands for *integer location*, so `iloc` only uses integer indexing to retrieve information from the data frames in the same way as indexing for Python lists.

## `iloc` for column selection? Think about it.
We can also access columns by position using `iloc` - but it is best not to if possible.

Example:

Suppose we want to access the 9th column in the df - then we want to select a column *by position*. In this case, the 9th column is the data at San Dieguito Wetland during spring and the 9th position corresponds to the index 8.

```python
# Select column by position using iloc
# The syntax is iloc[row-indices, column-indices]
# [:,8] means "select all rows from the column at index 8"
sdw_spring = df.iloc[:,8]
sdw_spring.head()
```

Unless you are *really* looking for information about the "first column", **do not access a column by position**. This can break in many ways:

- it relies on a person correctly counting the position of a column. Even with a small dataset this can be prone to error.

- it is not explicit: if we want information about SDW during spring, `df.SDW_spring` or `df['SDW_spring']` are explicitly telling us we are accessing that information. `df.iloc[:,8]` is obscure and uninformative.

- datasets can get updated. Maybe a new column was added before `SDW_spring`, changing the position of the column and making any code depending on `df.iloc[:,8]` invalid. 

**Accessing columns by labels helps reproducibility!**
