## Clipping 

**Clipping** means using a polygon (or polygons) to keep only the parts of the geometries that fall inside them.
the general syntax is:
```python
updated_geodf = geopandas.clip(geodf, mask)

```

where:
- `updated_geodf` = intersection of the geometries in `geodf` with `mask` 
- `geodf` = the `geopandas.GeoDataFrame` we want to clip
- `masks` = a `geopandas.GeoDataFrame` witht he polygons we will use for clipping

Rember `mask` and `geodf` must be in the same CRS.

````python
# clip populated to alask places
ak_places = gpd.clip(places, alaska)

fig, ax = plt.subplots()
alaska.plot(ax=ax)
alaska.plot(ax=ax)
ak_places.plot(ax=ax, color='red')
plt.show()
````

## Check CRS
You may have already noticed that the roads data is not in the same CRS as the alaska polygons, so these geo-datasets shouldn’t interact until they’re in the same CRS. Before jumping right into reprojecting and clipping, we will subset the data to select only US roads:

````python
# What is a road in this dataset
roads['type'].value_counts()
````

````python
# Filter by roads in usa, reduce tebular data before reducing via geometries
usa_roads = roads[roads['sov_a3'] == 'USA']
usa_roads.plot()
````

````python
# Clip USA roads to Alask geometry
ak_roads = gpd.clip(usa_roads.to_crs(alaska.crs), alaska)

fig, ax = plt.subplots()
alaska.plot(ax=ax)
ak_roads.plot(ax=ax, color='red')
plt.show()
````

## CLip with bounding box
To preserve roads:

We will clip the `usa_roads` geo-dataframe with the bounding box of `alaska` instead of its polygons. A bounding box is the smallest rectangle that contains all the geometries. It is defined by four coordinates, `minx, miny, maxx, maxy`:

````python

# Take a look at the values that define the bounding box
alaska.total_bounds
````

````python
# Reprojecting the roads into alaska crs
# using the toal bounds to clip

ak_complete_roads = gpd.clip(usa_roads.to_crs(alaska.crs), alaska.total_bounds)
````

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10,10))

ak_roads.plot(ax=ax1)
ax1.set_title('Roads clipped with AK multipolygon')

ak_complete_roads.plot(ax=ax2)
ax2.set_title('Roads clipped with bounding box')
````

## Measuring length 

Geodataframes habe a `length` attribute that calculates the length of each line in the geometry, **in the units of the CRS**.

Let's try it on one road in Alaska:

````python
# Reminder this is in ESPG 4326
usa_roads.crs
````

````python
# Ferry roads single line strings
usa_roads.loc[[1138]].length
````

````python
# reproject to the alaska CRS and measure length in kilometers
usa_roads.loc[[1138]].to_crs('epsg:3338').length / 1000
````

## How does clipping affect the data?

````python
# number of roads
print('Number of roads (polygon clip):', len(ak_roads))
print('Number of roads (bbox clip):', len(ak_complete_roads)) # no new roads same numbers

# Kilometer length
# Divide by 1000 to convert meters to kilometers
print('Total length in km (polygon clip):', round(ak_roads.length.sum() / 1000))
print('Total length in km (bbox clip):   ', round(ak_complete_roads.length.sum() / 1000)) # roads changed
````

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(2,1))

ak_roads.loc[[1138]].plot(ax=ax1)
ax1.set_title('Road 1138 clipped with AK multipolygon')

ak_complete_roads.loc[[1138]].plot(ax=ax2)
ax2.set_title('Road 1138 clipped with AK bounding box')
plt.show()
````

````python
# Why did this change? length_km, does not get modified only geometrylinestring to multilinestring
print('Length of 1138 in km(polygon clip):', round(ak_roads.loc[[1138]].length/1000 ))
print('Length of 1138 in km(bbox clip):', round(ak_complete_roads.loc[[1138]].length/1000))
````
